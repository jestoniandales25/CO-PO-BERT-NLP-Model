```python
import pdfplumber
import nltk
import re
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
# Extract Text Data from the Assigned Pages (5 pts)
with pdfplumber.open("Session 77 - 2022/77pv4.pdf") as pdf:
    text_content = ""
    for page in pdf.pages:
        text_content += page.extract_text() or ""
```

```python
# Display a Sample of the Extracted Text Data (5 pts)
print(text_content)
```

```python
# Perform Word-Level Tokenization (5 pts)
words = word_tokenize(text_content, language='english', preserve_line=False)
# Display the Number of Words (5 pts)
print(f"Number of words: {len(words)}")
```

```python
# Perform Sentence-Level Tokenization (5 pts)
sentences = sent_tokenize(text_content, language='english')
# Display the Number of Sentences (5 pts)
print(f"Number of sentences: {len(sentences)}")
```

```python
# Perform Paragraph-Level Tokenization
paragraphs = text_content.split('\n')
paragraphs = [p.strip() for p in paragraphs if p.strip()]
# Display the Number of Paragraphs (5 pts)
print(f"Paragraph tokenization complete. Total paragraphs extracted: {len(paragraphs)}")
```

```python

print(f"Total words: {len(words)}")
print(f"Total sentences: {len(sentences)}")
print(f"Total paragraphs: {len(paragraphs)}")
```
